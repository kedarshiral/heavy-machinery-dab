```python
# ============================================================
# config_setup — PROD ENVIRONMENT (single source of truth)
# ------------------------------------------------------------
# Har notebook '%run ../config_setup' se ye values uthata hai.
# DEPLOYMENT: sirf IS notebook ke values change karne se
# poora environment switch ho jata hai — koi aur notebook
# touch karne ki zaroorat nahi.
#
# STAGE se separation: alag schemas (bronze_prod/silver_prod/
# gold_prod/audit_prod) + alag checkpoints (_checkpoints_prod).
# Landing volume SHARED hai (same source files dono env padhte hain).
#
# NOTE (DAB): ye file repo ke MAIN branch me PROD values rakhti hai.
# stage branch pe isi file ke STAGE values hote hain.
# ============================================================

CATALOG = "main"

BRONZE_SCHEMA = "bronze_prod"
SILVER_SCHEMA = "silver_prod"
GOLD_SCHEMA = "gold_prod"
AUDIT_SCHEMA = "audit_prod"

# Landing volume — source files, checkpoints, ingestion config
HEAVY_SCHEMA = "heavy_machinery"
LANDING_VOLUME = "landing"
LANDING = f"/Volumes/{CATALOG}/{HEAVY_SCHEMA}/{LANDING_VOLUME}"

CHECKPOINTS = f"{LANDING}/_checkpoints_prod"
INGESTION_CONFIG_PATH = f"{LANDING}/_config/ingestion_config.json"

# Full schema identifiers — notebooks in f-strings use karte hain
BRONZE = f"{CATALOG}.{BRONZE_SCHEMA}"
SILVER = f"{CATALOG}.{SILVER_SCHEMA}"
GOLD = f"{CATALOG}.{GOLD_SCHEMA}"
AUDIT = f"{CATALOG}.{AUDIT_SCHEMA}"

# ============================================================
# Bootstrap — prod schemas agar nahi hain toh bana do (idempotent)
# ============================================================
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
for schema in [BRONZE_SCHEMA, SILVER_SCHEMA, GOLD_SCHEMA, AUDIT_SCHEMA]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}")

print(f"⚙️ config_setup [PROD] loaded → catalog={CATALOG} | bronze={BRONZE_SCHEMA} | silver={SILVER_SCHEMA} | gold={GOLD_SCHEMA}")
```
